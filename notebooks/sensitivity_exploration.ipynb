# Sensitivity of the 2026 and 2027 standards to the exempted-volume estimate

Exploratory view of the deterministic relationship between the additional 2025 exempted RVO and the total renewable fuel standards under the timing-rule family, using the transcribed inputs. Results used in the manuscript come from the pipeline, not from this notebook.

```python
from fractions import Fraction
import numpy as np
import matplotlib.pyplot as plt
from rfsrealloc.inputs import load_transcribed, set2_inputs, exempted_rvo_table
from rfsrealloc.reallocation import set2_srerv
from rfsrealloc.standards import obligated_volume

df = load_transcribed()
inp = set2_inputs(df)
stds, ervo = exempted_rvo_table(df)
base = set2_srerv(ervo)
den = {y: float(obligated_volume(inp[y])) for y in (2026, 2027)}
add = np.linspace(0, 1500, 16)  # million RINs of additional 2025 exempted RVO
for w, label in ((0.5, 'equal split'), (0.0, 'all in 2027')):
    s26 = 100 * (float(inp[2026].rfv['RF']) + (base[2026]['RF'] + w * add) / 1000) / den[2026]
    s27 = 100 * (float(inp[2027].rfv['RF']) + (base[2027]['RF'] + (1 - w) * add) / 1000) / den[2027]
    plt.plot(add, s26, label=f'2026, {label}')
    plt.plot(add, s27, '--', label=f'2027, {label}')
plt.axvline(770, color='k', lw=0.8)
plt.xlabel('additional 2025 exempted RVO (million RINs)')
plt.ylabel('total renewable fuel standard (%)')
plt.legend()
plt.show()
```
